# Outlier Detection 'Age'

```python
q1 = df1['Age'].quantile(0.25)
q3 = df1['Age'].quantile(0.75)
iqr = q3-q1
print(q1,q3,iqr)
```

```python
lower=q1-1.5*iqr
lower
```

```python
upper=q3+1.5*iqr
upper
```

```python
df[(df['Age'] < lower) | (df['Age'] > upper)]
```

```python
q1 = df1['Salary'].quantile(0.25)
q3 = df1['Salary'].quantile(0.75)
iqr = q3-q1
print(q1,q3,iqr)
```
